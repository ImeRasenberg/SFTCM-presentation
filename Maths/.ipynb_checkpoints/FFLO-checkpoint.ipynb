```python
# SageMath

var('k q xi_k Delta Delta_q muB H h omega')

# --- single-particle dispersions (Zeeman split) ---
xi_up   = xi_k - muB*H + i*h*omega
xi_down = xi_k + muB*H - i*h*omega

# ---------------------------
# 1. BCS Hamiltonian (q = 0)
# ---------------------------
G_0 = matrix([
    [xi_up, 0],
    [0, +xi_down]
])
Sigma = matrix([
    [0, Delta],
    [conjugate(Delta), 0]
])

H_BCS_matrix = G_0 + Sigma

show(H_BCS_matrix)

```
