```python
var('k q xi_k Delta muB H h omega beta V_0 V')

# --- 1. Define the base matrices ---
# G_0 is the inverse of the unperturbed single-particle Green's function matrix
# Note: In your snippet, G_1 was defined. Let's define G_0 properly as a diagonal matrix.
xi_up   = xi_k - muB*H + I*h*omega  # Use capital 'I' for the imaginary unit in Sage
xi_down = xi_k + muB*H - I*h*omega

G_0 = matrix([
    [1/xi_up, 0],
    [0, 1/xi_down]
])

# Pairing matrix (Sigma)
Sigma = matrix([
    [0, Delta],
    [conjugate(Delta), 0]
])

# --- 2. Function to compute the Trace of (Sigma * G_0)^n ---
def trace_order(n):
    # Compute the product matrix
    prod = (Sigma * G_0)^n
    # Take the trace and simplify
    tr = prod.trace()
    return tr

# --- 3. Calculate 2nd, 4th, and 6th orders ---
tr_2 = trace_order(2)
tr_4 = trace_order(4)
tr_6 = trace_order(6)

# --- 4. Display the results clearly ---
print("--- 0th Order Trace ---")
show(beta*V_0 * conjugate(Delta) *Delta/V)

print("--- 2nd Order Trace ---")
show(tr_2)

print("\n--- 4th Order Trace ---")
show(tr_4)

print("\n--- 6th Order Trace ---")
show(tr_6)
```
